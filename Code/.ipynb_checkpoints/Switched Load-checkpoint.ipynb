```python
import serial
import time

ser = serial.Serial('COM3', 115200)  # change port
time.sleep(2)  # wait for Pico reset
```

```python
while True:
    data = input()
    if data == "":
        break
    ser.write((data + "\n").encode())
    
    response = ser.readline()
    print(response.decode())
```
